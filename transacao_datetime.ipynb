```python
from datetime import datetime

def time_since_last_transaction(current_time, last_transaction_time):
    if last_transaction_time is None:
        return -1.0
    time_difference = current_time - last_transaction_time
    return time_difference.total_seconds()

t1 = datetime(2026, 1, 10, 12, 0, 0)
t2 = datetime(2026, 1, 10, 14, 30, 0)

# Retorna 9000.0 segundos (2h30min)
print(time_since_last_transaction(t2, t1))

# Primeira transação — retorna -1.0 (default)
print(time_since_last_transaction(t2, None))
```
